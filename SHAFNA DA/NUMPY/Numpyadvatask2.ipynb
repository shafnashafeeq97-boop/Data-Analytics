```python
# Question: Given the following NumPy array, perform the
# operations below.
#     import numpy as nparr = np.array([[12, 5, 18, 7],
#                                      [25, 10, 3, 20],
#                                      [8, 30, 15, 2]])
```

```python
import numpy as np
```

```python
# a. Display all elements greater than the overall mean of the array.
```

```python
a = np.array([[12, 5, 18, 7],
             [25, 10, 3, 20],
              [8, 30, 15, 2]])
print(np.mean(a))
print(a[a > 12])
```

```python
# b. Find the positions (row and column indices) of all elements divisible by 5.
```

```python
print(np.where(a % 5 == 0))
```

```python
print(np.argwhere(a % 5 == 0))
```

```python
# c. Replace all elements less than 10 with -1 without using a loop.
```

```python
a=np.where(a<10,-1,a)
print(a)
a[a<10]=-1
print(a)
```

```python
# d. Calculate the sum of each row and identify the row with the highest sum.
```

```python
b=np.sum(a,axis=1)
print(b)
print(np.max(b))
print(np.argmax(b))
```

```python
# e. Sort each row in ascending order and display the result.
```

```python
print(np.sort(a,axis=1))
```

```python
# f. Extract the last two columns and reshape them into a one-dimensional array.
```

```python
c=a[:,-2:]
print(c)
print(c.reshape(-1))
```

```python
# g. Create a flattened copy and a raveled view of the array. Modify one element in each and observe the effect on the original array.
```

```python
b=a.flatten()
b[0]=100
b=a.copy()
print(b)

c=a.ravel()
c[0]=100
c=a.view()
print(c)
```

```python
# h. Split the flattened array into three equal parts and display each part.
```

```python
a = np.array([[12, 5, 18, 7],
             [25, 10, 3, 20],
              [8, 30, 15, 2]])
print(a.flatten())
print(np.array_split(a,3))
```

```python
# i. Calculate the square root of all elements in the original array and round the results to two decimal places.
```

```python
print(np.round(np.sqrt(a),2))
```

```python
# j. Find the minimum and maximum values and their corresponding indices in the original array.
```

```python
print(np.min(a))
print(np.argmin(a))
print(np.max(a))
print(np.argmax(a))
```
