```python
# 17. Create an array containing [10, 20, 30, 40] . Create a separate copy and a view of the array.
# Modify the arrays and observe the effect on the original array.
```

```python
a=np.array([10, 20, 30, 40])
b=a.copy()
c=a.view()
print(b)
print(c)
b[0]=100
print(a)
print(b)
c[0]=100
print(a)
print(c)
```

```python
# 25. Create a multydimensional array. Create a flatten and ravel of the array.
# Modify the arrays and observe the effect on the original array.
```

```python
a=np.array([[[1,2,3,4]]])
b=a.flatten()
b[0]=100
print(a)
print(b)
c=a.ravel()
c[0]=100
print(a)
print(c)
```
