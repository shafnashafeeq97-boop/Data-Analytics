# <CENTER> Pandas Series

```python
import pandas as pd
```

### Creating a Series From a List:

```python
s=pd.Series([10,20,30,40,50])
print(s)
```

### Creating a Series From a List of Strings:

```python
s = pd.Series(["Apple", "Banana", "Mango", "Orange"])
print(s)
```

###  Series With a Custom Index:

```python
s = pd.Series([100, 200, 300, 400],index=["A", "B", "C", "D"])
print(s)
```

### Creating a Series From a Dictionary:

```python
data = {"A": 100,
        "B": 200,
        "C": 300,
        "D": 400}
s=pd.Series(data)
print(s)
```

### Accessing Series Elements - iloc:


```python
s=pd.Series([10,20,30,40,50])
print(s.iloc[2])
```

```python
print(s.iloc[4])
```

### loc - Label-Based Access:

```python
s = pd.Series([100, 200, 300, 400],index=["A", "B", "C", "D"])
print(s.loc["C"])
```

```python
print(s.loc["A"])
```

### Slicing a Series:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.iloc[:3])
```

```python
print(s.iloc[1:4])
```

### First and Last Elements:

```python
print(s.iloc[0])
```

```python
print(s.iloc[-1])
```

```python
# First three elements
print(s.iloc[:3])
```

```python
# Last two elements
print(s.iloc[-2:])
```

### Series Arithmetic Operations:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s+10)
```

```python
print(s-10)
```

```python
print(s*2)
```

```python
print(s/2)
```

### Mathematical Functions:


```python
marks = pd.Series([75, 88, 92, 65, 80])
print(marks.sum())
print(marks.mean()) 
print(marks.max()) 
print(marks.min())
```

### Important Statistical Methods:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.sum())
print(s.mean())
print(s.median())
print(s.max())
print(s.min())
```

### Filtering a Series:

```python
s = pd.Series([25, 60, 45, 80, 35, 70])
print(s[s>50])
```

```python
print(s[s<60])
```

### Checking Whether a Value Exists:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(30 in s.values)
print(100 in s.values)
```

```python
print(20 not in s.values)
print(60 not in s.values)
```
