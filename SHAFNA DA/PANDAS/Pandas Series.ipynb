# <CENTER> Pandas Series

```python
import pandas as pd
```

### Creating a Series From a List:

```python
s=pd.Series([10,20,30,40,50])
print(s)
```

### Creating a Series From a List of Strings:

```python
s = pd.Series(["Apple", "Banana", "Mango", "Orange"])
print(s)
```

###  Series With a Custom Index:

```python
s = pd.Series([100, 200, 300, 400],index=["A", "B", "C", "D"])
print(s)
```

### Creating a Series From a Dictionary:

```python
data = {"A": 100,
        "B": 200,
        "C": 300,
        "D": 400}
s=pd.Series(data)
print(s)
```

### Accessing Series Elements - iloc:


```python
s=pd.Series([10,20,30,40,50])
print(s.iloc[2])
```

```python
print(s.iloc[4])
```

### loc - Label-Based Access:

```python
s = pd.Series([100, 200, 300, 400],index=["A", "B", "C", "D"])
print(s.loc["C"])
```

```python
print(s.loc["A"])
```

### Slicing a Series:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.iloc[:3])
```

```python
print(s.iloc[1:4])
```

### First and Last Elements:

```python
print(s.iloc[0])
```

```python
print(s.iloc[-1])
```

```python
# First three elements
print(s.iloc[:3])
```

```python
# Last two elements
print(s.iloc[-2:])
```

### Series Arithmetic Operations:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s+10)
```

```python
print(s-10)
```

```python
print(s*2)
```

```python
print(s/2)
```

### Mathematical Functions:


```python
marks = pd.Series([75, 88, 92, 65, 80])
print(marks.sum())
print(marks.mean()) 
print(marks.max()) 
print(marks.min())
```

### Important Statistical Methods:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.sum())
print(s.mean())
print(s.median())
print(s.max())
print(s.min())
```

### Filtering a Series:

```python
s = pd.Series([25, 60, 45, 80, 35, 70])
print(s[s>50])
```

```python
print(s[s<60])
```

### Checking Whether a Value Exists:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(30 in s.values)
print(100 in s.values)
```

```python
print(20 not in s.values)
print(60 not in s.values)
```

### Unique Values:

```python
import pandas as pd
```

```python
s = pd.Series([10, 20, 10, 30, 20, 40])
print(s.unique())
```

### Number of Elements:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.size)
```

### Series Index:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.index)
```

### Series Values:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.values)
```

### Data Type - dtype:

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s.dtype)
```

### Adding a New Element:

```python
s = pd.Series([10, 20, 30], index=["A", "B", "C"])
s["D"] = 40
print(s)
```

### Modifying an Existing Element:

```python
s = pd.Series([10, 20, 30], index=["A", "B", "C"])
s["B"] = 200
print(s)
```

### Dropping an Element:

```python
s = pd.Series([10, 20, 30, 40],index=["A", "B", "C", "D"])
s = s.drop("C")
print(s)
```

### Sorting a Series - Ascending Order:

```python
s = pd.Series([25, 10, 45, 30, 15])
print(s.sort_values())
```

### Sorting a Series - Descending Order:

```python
s = pd.Series([25, 10, 45, 30, 15])
print(s.sort_values(ascending=False))
```

### Resetting the Index After Sorting:

```python
s = pd.Series([25, 10, 45, 30, 15])
result = s.sort_values().reset_index(drop=True)
print(result)
```

### Missing Values in a Series:

```python
import numpy as np
s = pd.Series([10, 20, np.nan, 40, np.nan])
print(s)
```

### Checking for Missing Values - isna():

```python
print(s.isna())
```

### Removing Missing Values - dropna():

```python
s = pd.Series([10, 20, np.nan, 40, np.nan])
print(s.dropna())
```

### Filling Missing Values - fillna():

```python
s = pd.Series([10, 20, np.nan, 40])
s = s.fillna(0)
print(s)
```

### Adding Two Series Together:

```python
s1 = pd.Series([10, 20, 30])
s2 = pd.Series([5, 10, 15])
print(s1 + s2)
```

```python
print(s1-s2)
```

```python
print(s1*s2)
```

```python
print(s1/s2)
```

### head() and tail():

```python
s = pd.Series([10, 20, 30, 40, 50, 60, 70])
print(s.head())  
print(s.tail())  
```

```python
print(s.head(3)) 
print(s.tail(2)) 
```

### Series vs a Plain List:

```python
numbers = [10, 20, 30, 40] # a plain list
numbers = pd.Series([10, 20, 30, 40]) 
print(numbers.mean())
print(numbers.max())
print(numbers.min())
print(numbers.sort_values())
```
