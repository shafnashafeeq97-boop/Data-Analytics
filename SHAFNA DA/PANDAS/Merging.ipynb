# <CENTER> Pandas Data Integration and Transformation

```python
import pandas as pd
```

### Basic Merging

```python
df1=pd.DataFrame({
    "id":[1,2,3,4,5],
    "name":["Anu","Aju","Amal","Meera","Riya"],
    "age":[20,23,21,22,24]
})    
df2=pd.DataFrame({
    "id":[1,2,6,4,9],
    "mark":[50,65,70,60,80],
    "grade":["A+","A","B","A+","A"]
})
df=pd.merge(df1,df2, on="id")
print(df)    
```

### inner join(default)

```python
pd.merge(df1,df2, on="id", how="inner")
```

### Left join

```python
pd.merge(df1,df2, on="id" , how="left")
```

### Right join

```python
pd.merge(df1,df2, on="id", how="right")
```

### Outer join

```python
pd.merge(df1,df2, on="id", how="outer")
```

### Merge on Different Column Names

```python
df1=pd.DataFrame({
    "id":[1,2,3,4,5],
    "name":["Anu","Aju","Amal","Meera","Riya"],
    "age":[20,23,21,22,24]
})    
df2=pd.DataFrame({
    "std_id":[1,2,6,4,9],
    "mark":[50,65,70,60,80],
    "grade":["A+","A","B","A+","A"]
})
df=pd.merge(df1,df2, left_on="id",right_on="std_id")
print(df)    
```

```python
merge1=pd.merge(df1,df2,left_on="id",right_on="std_id",how="outer")
print(merge1)
```

```python
merge2=pd.merge(df1,df2,left_on="id",right_on="std_id",how="left")
print(merge2)
```

```python
merge3=pd.merge(df1,df2,left_on="id",right_on="std_id",how="right")
print(merge3)
```

```python
merge4=pd.merge(df1,df2,left_on="id",right_on="std_id",how="outer")
print(merge4)
```

### concatnation

```python
pd.concat([df1,df2], axis=0) 
```

```python
pd.concat([df1,df2], axis=1) 
```
