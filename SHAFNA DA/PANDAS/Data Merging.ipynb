## MERGING DATA USING PANDAS

### Basic Merging:

```python
import pandas as pd
```

```python
df1=pd.DataFrame({"id":[1,2,3],
                  "Name":["Aisha","Binu","Cathy"]})
df2=pd.DataFrame({"id":[1,2,4],
                  "Mark":[85,90,75]})
merged_df=pd.merge(df1,df2,on="id")
print(merged_df)
```

## <center> TYPES OF JOINS

## INNER JOIN(Default)

### Only Common Rows:

```python
pd.merge(df1,df2,on="id",how="inner")
```

## LEFT JOIN

### All Rows from Left DataFrame And Common from Right:

```python
pd.merge(df1,df2,on="id",how="left")
```

## RIGHT JOIN

### All Rows from Right DataFrame And Common from Left:

```python
pd.merge(df1,df2,on="id",how="right")
```

## OUTER JOIN

### All Rows from Both:

```python
pd.merge(df1,df2,on="id",how="outer")
```

## <CENTER> MERGE ON DIFFERENT COLUMN NAMES:

```python
df1=pd.DataFrame({"id":[1,2,3],
                  "Name":["Aisha","Binu","Cathy"]})
df2=pd.DataFrame({"student_id":[1,2,4],
                  "Mark":[85,90,75]})
merged_df=pd.merge(df1,df2,left_on="id",right_on="student_id")
print(merged_df)
```

```python
l_merge=pd.merge(df1,df2,left_on="id",right_on="student_id",how="left")
l_merge
```

```python
r_merge=pd.merge(df1,df2,left_on="id",right_on="student_id",how="right")
r_merge
```

```python
o_merge=pd.merge(df1,df2,left_on="id",right_on="student_id",how="outer")
o_merge
```

## <CENTER> CONCATENATION

```python
pd.concat([df1,df2],axis=0) 
```

```python
pd.concat([df1,df2],axis=1)
```

## <CENTER> APPLY

```python
Marks=pd.Series([45,60,75,90])
def grade(x):
    if x>=50:
        return "Pass"
    else:
        return "Fail"
print(Marks.apply(grade))        
```

### Apply Using Lambda Function:

```python
Marks=pd.Series([45,60,75,90])
result=Marks.apply(lambda x:x+5)
print(result)
```

### Apply() On DataFrame Columns:

```python
df=pd.DataFrame({"Math":[80,70,90],
                 "Science":[85,60,95]})
print(df.apply(max))
```

```python
print(df.apply(min))
```
