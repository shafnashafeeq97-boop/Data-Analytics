<h3 style="color:#1F4E78; margin-top:0;">DATA CLEANING</h3>

```python
df["Gender"] = df["Gender"].replace({
    "F": "Female",
    "female": "Female",
    "M": "Male",
    "male": "Male"})
df["Gender"] 
```

```python
df["Age"]=df["Age"].replace(200,0,inplace=True)
df["Age"]
```

```python
df["Age"]=df["Age"].replace(-5,0,inplace=True)
df["Age"]
```

```python
df["purchase_amount"]=df["purchase_amount"].replace(-999,0,inplace=True)
df["purchase_amount"]
```

```python
df["feedback_score"]=df["feedback_score"].replace(-1,0,inplace=True)
df["feedback_score"]
```

```python
df["CustomerID"]=df["CustomerID"].fillna(df["CustomerID"].mode()[0] ,inplace=True)
df["CustomerID"]
```

```python
df["Name"]=df["Name"].fillna("Unknown", inplace=True)
df["Name"]
```

```python
df["Gender"]=df["Gender"].fillna("Unknown",inplace=True)
df["Gender"]
```

```python
df["Age"]=df["Age"].fillna(df["Age"].median(), inplace=True)
df["Age"]
```

```python
df["City"]=df["City"].fillna("Unknown",inplace=True)
df["City"]
```

```python
df["Signup_Date"]=df["Signup_Date"].fillna(df["Signup_Date"].mode()[0],inplace=True)
df["Signup_Date"]
```

```python
df["Last_purchase_date"]=df["Last_purchase_date"].fillna(df["Last_purchase_date"].mode()[0],inplace=True)
df["Last_purchase_date"]
```

```python
df["purchase_amount"]=df["purchase_amount"].fillna(df["purchase_amount"].median(),inplace=True)
df["purchase_amount"]
```

```python
df["feedback_score"]=df["feedback_score"].fillna(df["feedback_score"].mean(),inplace=True)  
df["feedback_score"]
```

```python
df["email"]=df["email"].fillna("Unknown",inplace=True)
df["email"]
```

```python
df["Phone_number"]=df["Phone_number"].fillna(df["Phone_number"].mode()[0],inplace=True)
df["Phone_number"]
```

```python
df["Country"]=df["Country"].fillna("Unknown",inplace=True)
df["Country"]
```

```python
df.head()
```

```python
df.tail()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

#### 💡 Insight

- *Unknown:* Used to fill missing categorical values.
- *Mode:* Filled missing values with the most frequent value.
- *Mean:* Filled missing numerical values with the average.
- *Median:* Filled missing numerical values with the middle value.

*Conclusion:* Different methods were used to handle missing values and improve data completeness.
