## Data Selection Using loc and iloc

### loc (Label-Based Indexing):

```python
import pandas as pd
```

```python
data= {
"Name": ["Asha","Rahul","Anu"],
"Age": [20,21,19],
"City": ["Kochi","Delhi","Chennai"]
}
d=pd.DataFrame(data)
print(d)
```

```python
print(d.loc[1])
```

```python
print(d.loc[0:1, ["Name","Age"]])
```

### iloc (Integer Position-Based Indexing):

```python
print(d.iloc[1])
```

```python
print(d.iloc[0:2][0:2])
```

### Difference in Slicing:

#### Using loc

```python
print(d.loc[0:1])
```

#### Using iloc

```python
print(d.iloc[0:1])
```

### Selecting a Single Value:

#### loc

```python
print(d.loc[2,"City"])
```

```python
print(d.loc[0,"Age"])
```

#### iloc

```python
print(d.iloc[2,2])
```

```python
print(d.iloc[0,0])
```
