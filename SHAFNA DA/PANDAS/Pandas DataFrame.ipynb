# <center> Pandas DataFrame

```python
import pandas as pd
```

### Creating a DataFrame from a Dictionary:

```python
data = [["Anu", 21, 85],
        ["Rahul", 22, 90],
        ["Priya", 20, 88]]
df = pd.DataFrame(data)
print(df)
```

```python
print(df.ndim)
```

### Creating a DataFrame from a List:

```python
data = [["Anu", 21, 85],
        ["Rahul", 22, 90],
        ["Priya", 20, 88]]
df = pd.DataFrame(data, columns=["Name", "Age", "Mark"])
print(df)
```

### Creating a DataFrame with a Custom Index:

```python
data = {"Name": ["Anu", "Rahul", "Priya"],
        "Mark": [85, 90, 88]}
df = pd.DataFrame(data, index=["S1", "S2", "S3"])
print(df)
```

### Selecting Rows Using loc:

```python
data = {"Name": ["Anu", "Rahul", "Priya"],
        "Mark": [85, 90, 88]}
df = pd.DataFrame(data, index=["S1", "S2", "S3"])
print(df.loc["S2"])
```

### Adding a New Column:

```python
df = pd.DataFrame({"Name": ["Anu", "Rahul", "Priya"],
                   "Mark": [85, 90, 88]})
df["Grade"]=["A","A+","A"]
print(df)
```

### Modifying an Existing Column:

```python
df["Marks"]=[90,95,92]
print(df)    
```

### Adding a New Row:

```python
df = pd.DataFrame({"Name": ["Anu", "Rahul", "Priya"],
                   "Mark": [85, 90, 88]})
df.loc[3] = ["Arun", 95]
print(df)
```

### Removing a Column - drop():

```python
df = df.drop("Mark", axis=1)
print(df)
```

### Removing a Row - drop():

```python
df = pd.DataFrame({"Name": ["Anu", "Rahul", "Priya"],
                   "Mark": [85, 90, 88]})
df = df.drop(1)
print(df)
```

### Basic Information - info():

```python
df = pd.DataFrame({"Mark": [70, 80, 90, 85, 75]})
df.info()
```

### Basic Statistics - describe():

### Filtering Rows:

```python
df = pd.DataFrame({"Name": ["Anu", "Rahul", "Priya", "Arun"],
                   "Mark": [85, 90, 45, 75]})
print(df[df["Mark"] > 80])
```

```python
print(df[df["Mark"]<85])
```

### Sorting a DataFrame:

```python
# Ascending order
df = df.sort_values("Mark")
print(df)
```

```python
df = df.sort_values("Name")
print(df)
```

```python
# Descending order
df = df.sort_values("Mark", ascending=False)
print(df)
```

### Checking for Missing Values - isna():

```python
import numpy as np
df = pd.DataFrame({"Name": ["Anu", "Rahul", "Priya"],
                   "Mark": [85, np.nan, 90]})
print(df.isna())
```

### Counting Missing Values:

```python
print(df.isna().sum())
```

### Filling Missing Values - fillna():

```python
df["Mark"] = df["Mark"].fillna(0)
print(df)
```

### Removing Rows with Missing Values - dropna():

```python
df=df.dropna()
print(df)
```

### Viewing First and Last Rows - head() and tail():

```python
print(df.head()) 
print(df.tail()) 
```

```python
print(df.head(3)) 
print(df.tail(2))
```
